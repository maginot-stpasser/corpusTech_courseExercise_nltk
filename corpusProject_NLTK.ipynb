```python
import nltk
from nltk.corpus import stopwords
import string
from nltk.text import Text
from nltk.collocations import *

path = "/content/drive/MyDrive/Colab Notebooks/corpus_project/input_speech.txt"
dictA = {}
text_tok = []
stop = stopwords.words("english") + list(string.punctuation) + ["''", "``"]
# some unusual punctuations are generated after being passed to word tokeniser



with open(path, "r", encoding='utf-8') as f:
    for line in f:

        # there are unusual punctuations "—" in the text, not a very efficient way to eliminate it
        temp = line.split("—")
        lineTemp = ""
        for items in temp:
            lineTemp += items

        sent = nltk.tokenize.sent_tokenize(lineTemp, language='english')    # tokenise the lines into sent
        for items in sent:
            for word in nltk.word_tokenize(items, language="english"):    # tokenise the sentences into words
                temp = word.lower()
                text_tok.append(word)   # appending words to the token list with capitalisation kept
                if temp not in dictA:
                    dictA[temp] = 1     # add words (no stop words and punc) into dictionary without capitalisation
                    # print(temp)
                else:
                    dictA[temp] += 1    # subsequent counting


print("\nTokenised text: ", "\n", text_tok, "\n")     # tokenised text


dictA_sorted = dict(sorted(dictA.items(), key = lambda x:x[1], reverse=True))
# print(dictA_sorted)   # a sorted dictionary (with stopwords and punc)
dictB = dictA_sorted.copy()   # using the .copy() method to create a new dictionary without referencing back to dictA_sorted
dictC = dictA_sorted.copy()


for items in list(string.punctuation) + ["''", "``"]:
    dictB.pop(items, None)
# print(dictB)          # sorted dictionary (with stopwords)


for items in stop:
    dictC.pop(items, None)
print("\nThe frequency list (without stopwords and punctuations):\n", dictC, "\n")
# sorted dictionary (without stopwords and punc), i.e., a clean frequency list in form of a sorted dictionary


# finding concordances
textCorpus = Text(text_tok)   # processed with Text method in order to be passed to the concordance method
targetWords = ["way", "came", "election", "democrat"]
print(f"\nTarget words: {targetWords}")
print("\nConcordances:")
for word in targetWords:
    textCorpus.concordance(word, lines=10, width=80)    # lines=10 (displaying 10 matches), width=80 (default)
    print()

# finding Bigrams (collocations) with the target words
colloList = []
Bigram_measures = nltk.collocations.BigramAssocMeasures()
lower_textTok = list(x.lower() for x in text_tok)
lowerTextCorp = Text(lower_textTok)
# the bigram filter does not recognise capitalised letters, using (x.lower() for x in...)
# to turn the tokens and the target words into lower case
print("\nBigram Collocations (top 5 pmi score):\n")
for word in (x.lower() for x in targetWords):
    bi_finder = BigramCollocationFinder.from_words(lowerTextCorp)
    my_filter = lambda *w: word not in w    # filtering the bigrams without the target words
    bi_finder.apply_ngram_filter(my_filter)
    # for punc in list(string.punctuation):     # in case there is a need to filter punctuations
    #     my_filter = lambda *w: punc in w
    #     bi_finder.apply_ngram_filter(my_filter)
    colloList.append(bi_finder.nbest(Bigram_measures.pmi, 5))   # append the results with the top 5 pmi score into a list
for items in colloList:
    print(items)
print()

colloListB = []
Trigram_measures = nltk.collocations.TrigramAssocMeasures()
lower_textTok = list(x.lower() for x in text_tok)
lowerTextCorp = Text(lower_textTok)
# the bigram filter does not recognise capitalised letters, using (x.lower() for x in...)
# to turn the tokens and the target words into lower case
print("\nTrigram Collocations (top 5 pmi score):\n")
for word in (x.lower() for x in targetWords):
    tri_finder = TrigramCollocationFinder.from_words(lowerTextCorp)
    my_filter = lambda *w: word not in w    # filtering the bigrams without the target words
    tri_finder.apply_ngram_filter(my_filter)
    # for punc in list(string.punctuation):     # in case there is a need to filter punctuations
    #     my_filter = lambda *w: punc in w
    #     bi_finder.apply_ngram_filter(my_filter)
    colloListB.append(tri_finder.nbest(Trigram_measures.pmi, 5))   # append the results with the top 5 pmi score into a list
for items in colloListB:
    print(items)
print()

# finding total number of words (with stopwords)
counter = 0
for items in dictB.values():
    counter += items
print("\nTotal number of words (not tokens):\n", counter, "\n")

# finding vocab size
print("\nVocabulary size:\n", len(dictB.keys()), "\n")    # vocabulary size (with stopwords and without punc)

###############################
# # When tokens with punctuations
# # type token ratio
# ttRatio = len(dictA_sorted.keys()) / len(text_tok)
# print("\nType-Token Ratio (tokens, not words): \n", ttRatio, "\n")

# # content word ratio
# print("\nContent word ratio (content words/tokens):\n", len(dictC.keys()) / len(text_tok), "\n")

###############################

ttRatio = len(dictB.keys()) / counter
print("\nType-Token (Type-Word) Ratio (number of types of words[==vocabulary size]/number of words): \n", ttRatio, "\n")

# content word ratio
cwRatio = len(dictC.keys()) / counter
print("\nContent word ratio (number of types of content words/number of words):\n", cwRatio, "\n")
```
